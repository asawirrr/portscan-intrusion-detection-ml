```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler


# Load first file
df1 = pd.read_csv('Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv')

# Load second file
df2 = pd.read_csv('Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv')

# Load 3rd file
df3 = pd.read_csv('Friday-WorkingHours-Morning.pcap_ISCX.csv')


# Load 4th file
df4 = pd.read_csv('Monday-WorkingHours.pcap_ISCX.csv')


# Load 5th file
df5 = pd.read_csv('Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv')


# Load 6th file
df6 = pd.read_csv('Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv')

# Load 7th file
df7 = pd.read_csv('Tuesday-WorkingHours.pcap_ISCX.csv')

# Load 8th file
df8 = pd.read_csv('Wednesday-workingHours.pcap_ISCX.csv')


# List of all dataframes
dfs = [df1, df2, df3, df4,df5, df6, df7, df8]

# Merge all dataframes vertically (along rows)
merged_df = pd.concat(dfs, ignore_index=True)

# Check the merged dataframe
print(merged_df.columns)
# Now, you can check the class distribution
class_counts = merged_df[' Label'].value_counts()
print("Class distribution:\n", class_counts)
print(merged_df.shape)


```

```python
# Strip leading/trailing spaces from column names
merged_df.columns = merged_df.columns.str.strip()

print(merged_df.columns)
print(merged_df.shape)

```

```python
# Show all rows in the output
pd.set_option('display.max_rows', None)
# Filter only BENIGN rows
benign_df = merged_df[merged_df['Label'] == 'BENIGN']  # Assuming 0 = BENIGN

# Count the destination port values
#print(benign_df['Destination Port'].value_counts())
import warnings
warnings.filterwarnings("ignore")  
```

```python
# Filter only BENIGN and PortScan
filtered_df = merged_df[merged_df['Label'].isin(['BENIGN', 'PortScan'])].copy()
print('\n')
print(filtered_df.shape)
print(filtered_df['Label'].value_counts())

# Drop single-value (constant) columns
filtered_df = filtered_df.loc[:, filtered_df.nunique() > 1]
print('\n')
print(filtered_df.shape)
print(filtered_df['Label'].value_counts())

# Replace infinite values with NaN
filtered_df.replace([np.inf, -np.inf], np.nan, inplace=True)
filtered_df.dropna(inplace=True)
print('\n')
print(filtered_df.shape)
print(filtered_df['Label'].value_counts())

# Label encoding (BENIGN → 0, PortScan → 1)
filtered_df['Label'] = filtered_df['Label'].replace({'BENIGN': 0, 'PortScan': 1})
print('\n')
print(filtered_df.shape)
print(filtered_df['Label'].value_counts())



# Sample 50,000 from each class AFTER preprocessing
df_benign = filtered_df[filtered_df['Label'] == 0].sample(n=50000, random_state=42)
df_portscan = filtered_df[filtered_df['Label'] == 1].sample(n=50000, random_state=42)
df = pd.concat([df_benign, df_portscan], ignore_index=True).sample(frac=1, random_state=42).reset_index(drop=True)
print('\n')
print(df.shape)
print(df['Label'].value_counts())
```

```python
# Split features/labels
X = df.drop(columns=['Label'])
y = df['Label']

# Standardize numeric features
X = pd.DataFrame(StandardScaler().fit_transform(X), columns=X.columns)

```

```python
from sklearn.feature_selection import SelectKBest, f_classif

# Select top 20 features
selector = SelectKBest(score_func=f_classif, k=20)
X_new = selector.fit_transform(X, y)
selected_features = X.columns[selector.get_support()]

# Update X_train and X_test with selected features
X = pd.DataFrame(X_new, columns=selected_features)
print("Selected Features:", selected_features)

# Train-test split (with selected features)
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)
```

```python
# Feature Importance Plot
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

plt.figure(figsize=(10,6))
pd.Series(rf.feature_importances_, index=selected_features).sort_values().plot(kind='barh')
plt.title("Top 20 Features Importance (Random Forest)")
plt.show()
```

```python
from sklearn.metrics import classification_report, confusion_matrix

models = {
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(probability=True, random_state=42),
    "FNN": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=300, random_state=42)
}

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)  

    results[name] = {
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1-score": f1,
        "Confusion Matrix": cm        
    }

    emoji = "🔍"
    full_name = {
        "Random Forest": "Random Forest",
        "SVM": "Support Vector Machine",
        "FNN": "Neural Network"
    }[name]

    print(f"\n{emoji} {full_name} Evaluation:")
    print(f"Accuracy: {acc}")
    print(f"Precision: {prec}")
    print(f"Recall: {rec}")
    print(f"F1-Score: {f1}")
```

```python
for name in models:
    cm = results[name]['Confusion Matrix']
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f"{name} - Confusion Matrix")
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```

```python
# Extract metrics
metrics = []
for model_name in results:
    metrics.append({
        'Model': model_name,
        'Accuracy': results[model_name]['Accuracy'],
        'Precision': results[model_name]['Precision'],
        'Recall': results[model_name]['Recall'],
        'F1-Score': results[model_name]['F1-score']
    })

# Plot
metrics_df = pd.DataFrame(metrics).set_index('Model')
metrics_df.plot(kind='bar', figsize=(12, 6))
plt.title("Model Performance Comparison")
plt.ylim(0.9, 1.0)
plt.legend(bbox_to_anchor=(1.05, 1))
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

for name, model in models.items():
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:  # SVM case
        y_prob = model.decision_function(X_test)
    
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f'{name} (AUC = {roc_auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--')  # Diagonal line
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve (PortScan Detection)')
plt.legend()
plt.show()
```

```python
cv_scores = cross_val_score(RandomForestClassifier(random_state=42), X, y, cv=5, scoring='accuracy')
print(f"RF 5-Fold CV Accuracy: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
```
